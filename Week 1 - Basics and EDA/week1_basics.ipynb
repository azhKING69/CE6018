```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("todo.csv", encoding="cp1252")
```

```python
# Earthquake Magnitude
df2 = df[["Earthquake_Magnitude"]]
df2 = df2[df2["Earthquake_Magnitude"] >= 4]


plt.hist(df2["Earthquake_Magnitude"], bins=50)
plt.xlabel("Earthquake Magnitude")
plt.ylabel("Frequency")
plt.title("Histogram of Earthquake Magnitudes")
plt.show()
```

```python
stats2 = {
    "Mean": df2.mean(),
    "Median": df2.median(),
    "Standard Deviation": df2.std(),
    "Skewness": df2.skew(),
    "Kurtosis": df2.kurt()
}

print("Earthquake Magnitude Statistics")
print("Mean:", stats2["Mean"].values)
print("Median:", stats2["Median"].values)
print("Standard Deviation:", stats2["Standard Deviation"].values)
print("Skewness:", stats2["Skewness"].values)
print("Kurtosis:", stats2["Kurtosis"].values)
```
